```python
import matplotlib.pyplot as plt
import numpy as np
import math 
```

```python
data_file = open('./maze_graph.txt', 'r')
n2,  = [ int(x) for x in data_file.readline().split() ]

n = int (math.sqrt(n2))
print ('maze size : %d x %d' % (n, n))

neib_num = [0] * n2
neib_vec = [[]] * n2
p_vec = [[]] * n2
for idx in range(n2):
    line = data_file.readline()
    neib_num[idx] = int(line)
    
    line = data_file.readline()
    darray = [float (x) for x in line.split()] 
    neib_vec[idx] = [int (x) for x in darray[::2]]
    p_vec[idx] = darray[1::2]
data_file.close()
```

```python
reached = [False] * n2

reached[0] = True

parent = [-1] * n2

new_sites = [0]

p = 0
while p < len(new_sites):
    idx = new_sites[p]
    for i in range(neib_num[idx]):    
        idx_new = neib_vec[idx][i]
        if reached[idx_new] is False:
            new_sites.append(idx_new)
            parent[idx_new] = idx
            reached[idx_new] = True
    p = p + 1
    if reached[n2-1] :
        break

y = n2-1    
path = [y]

while y != 0:
    y = parent[y]
    path.append(y)

print (path)
```

```python
fig = plt.figure(figsize=(9.0,8.5))
ax = plt.gca() 

data_file = open('./g.txt', 'r')
n2,  = [ int(x) for x in data_file.readline().split() ]

n = int (math.sqrt(n2))

print ('maze size : %d x %d' % (n, n))

darray = []
offsets = [[0,1], [0,-1], [-1,0], [1,0]]

for idx in range(n2):

    x = idx % n 
    y = idx // n 

    for i in range(4) :
        xx = x + offsets[i][0]
        yy = y + offsets[i][1]

        if (0 <= xx <= n-1) and (0 <= yy <= n-1) :
            idx1 = yy * n + xx 

            if (idx1 not in neib_vec[idx]) and (idx1 > idx) :
                if xx == x :
                    ymin = min(y, yy) + 0.5 
                    plt.hlines(ymin, x-0.5, x+0.5) 
                else : 
                    xmin = min(x, xx) + 0.5 
                    plt.vlines(xmin, y-0.5, y+0.5) 
   
plt.vlines(-0.5, 0.5, n-0.5) 
plt.vlines(n-0.5, -0.5, n-1.5) 
plt.hlines(-0.5, -0.5, n-0.5) 
plt.hlines(n-0.5, -0.5, n-0.5) 

plt.text(-0.3,-0.4, 'A', fontsize=9) 
plt.text(n-1.3,n-1.3, 'B', fontsize=9) 
plt.arrow(-1.4,0.0, 0.6, 0, head_width=0.5, head_length=0.2)
plt.arrow(n-0.2,n-1, 0.6, 0, head_width=0.5, head_length=0.2)


pts = [divmod(x, n) for x in path[::-1]]
for i in range(0, len(path)-1) :
  if pts[i][1] == pts[i+1][1] :
    plt.vlines(pts[i][1], pts[i][0], pts[i+1][0], linewidth=1.5, linestyle='--',color='k') 
  else : 
    plt.hlines(pts[i][0], pts[i][1], pts[i+1][1], linewidth=1.5, linestyle='--', color='k') 
    
fig.patch.set_visible(False)
ax.axis('off')

for tic in ax.yaxis.get_major_ticks():
    tic.tick1On = tic.tick2On = False

for tic in ax.xaxis.get_major_ticks():
    tic.tick1On = tic.tick2On = False

ax.set_xlim( [-2, n+1] )
ax.set_ylim( [-1, n] )
ax.set_xticklabels([])
ax.set_yticklabels([])

plt.show()
```

```python
mat = np.zeros((n2, n2))
b = np.zeros(n2)
print (mat.shape)
```

```python
for idx in range(n2):
    if idx == 0 :
        mat[idx][idx] = 1.0
        b[idx]=0
    if idx == n2-1 :
        mat[idx][idx] = 1.0
        b[idx]=1
    if idx > 0 and idx < n2 - 1:
        for i in range(neib_num[idx]):
            idx1 = neib_vec[idx][i]
            mat[idx][idx1] = p_vec[idx][i]
        mat[idx][idx] = -1
        b[idx] = 0

q = np.linalg.solve(mat, b)        
```
